```python
import pandas as pd
from numpy.random.mtrand import multinomial
from sklearn.model_selection import train_test_split as tts
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv(r"C:\Users\LENOVO\Desktop\aio_learn_csv\snappfood.csv")

df.head(10)
```

```python
df.drop(["comment", "comment_length"], axis=1, inplace=True)
```

```python
df.head()
```

```python
x = df["comment_cleaned"]
y = df["label"]
```

```python
x_train, x_test, y_train, y_test = tts(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
vectorizer = CountVectorizer()
x_train = vectorizer.fit_transform(x_train)
x_test = vectorizer.transform(x_test)
```

```python
model = MultinomialNB()
model.fit(x_train, y_train)
```

```python
pred = model.predict(x_test)
accuracy = accuracy_score(y_test, pred)
print(accuracy)
```

```python

import tkinter as tk
from tkinter import ttk, messagebox

# ================= COLORS =================

BG = "#101212"
PANEL = "#191D1C"
CARD = "#202624"
GREEN = "#00E58B"
GREEN_DARK = "#00B96F"
WHITE = "#F1F5F3"
GRAY = "#929D98"
BORDER = "#303936"
RED = "#FF6565"

# ================= WINDOW =================

root = tk.Tk()
root.title("SNAPPFOOD | AI FEEDBACK")
root.geometry("800x620")
root.minsize(650, 520)
root.configure(bg=BG)

# ================= STYLE =================

style = ttk.Style()
style.theme_use("clam")

style.configure(
    "Modern.TButton",
    font=("Segoe UI", 10, "bold"),
    background=GREEN,
    foreground="#07130D",
    padding=(16, 10),
    borderwidth=0
)

style.map(
    "Modern.TButton",
    background=[("active", GREEN_DARK)]
)

# ================= CUSTOM SCROLLBAR =================

style.configure(
    "Dark.Vertical.TScrollbar",
    background=CARD,
    troughcolor=PANEL,
    bordercolor=PANEL,
    arrowcolor=GREEN,
    lightcolor=CARD,
    darkcolor=CARD,
    gripcount=0,
    width=8
)

style.map(
    "Dark.Vertical.TScrollbar",
    background=[
        ("active", GREEN_DARK),
        ("!active", CARD)
    ]
)

# ================= HEADER =================

header = tk.Frame(root, bg=BG)
header.pack(fill="x", padx=28, pady=(18, 12))

brand = tk.Frame(header, bg=BG)
brand.pack(side="left")

tk.Label(
    brand, text="SNAPP",
    font=("Segoe UI", 21, "bold"),
    fg=WHITE, bg=BG
).pack(side="left")

tk.Label(
    brand, text="FOOD",
    font=("Segoe UI", 21, "bold"),
    fg=GREEN, bg=BG
).pack(side="left")

tk.Label(
    header,
    text="● AI FEEDBACK SYSTEM ONLINE",
    font=("Segoe UI", 8, "bold"),
    fg=GREEN, bg=BG
).pack(side="right", pady=10)

tk.Frame(root, bg=BORDER, height=1).pack(fill="x", padx=28)

# ================= MAIN =================

main = tk.Frame(root, bg=BG)
main.pack(fill="both", expand=True, padx=28, pady=17)

tk.Label(
    main,
    text="Customer Feedback",
    font=("Segoe UI", 21, "bold"),
    fg=WHITE, bg=BG
).pack(anchor="w")

tk.Label(
    main,
    text="Share your experience with SnappFood.",
    font=("Segoe UI", 9),
    fg=GRAY, bg=BG
).pack(anchor="w", pady=(3, 12))

# ================= CHAT PANEL =================

chat_panel = tk.Frame(
    main,
    bg=PANEL,
    highlightbackground=BORDER,
    highlightthickness=1
)
chat_panel.pack(fill="both", expand=True)

chat_header = tk.Frame(chat_panel, bg=PANEL, padx=15, pady=11)
chat_header.pack(fill="x")

tk.Label(
    chat_header,
    text="SNAPPFOOD AI ASSISTANT",
    font=("Segoe UI", 9, "bold"),
    fg=WHITE, bg=PANEL
).pack(side="left")

tk.Label(
    chat_header,
    text="● ACTIVE",
    font=("Segoe UI", 8, "bold"),
    fg=GREEN, bg=PANEL
).pack(side="right")

tk.Frame(chat_panel, bg=BORDER, height=1).pack(fill="x")

# ================= CHAT HISTORY =================

chat_container = tk.Frame(chat_panel, bg=PANEL)
chat_container.pack(fill="both", expand=True, padx=12, pady=10)

chat_canvas = tk.Canvas(
    chat_container,
    bg=PANEL,
    highlightthickness=0
)

scrollbar = ttk.Scrollbar(
    chat_container,
    orient="vertical",
    command=chat_canvas.yview,
    style="Dark.Vertical.TScrollbar"
)

chat_canvas.configure(yscrollcommand=scrollbar.set)

scrollbar.pack(side="right", fill="y")
chat_canvas.pack(side="left", fill="both", expand=True)

chat_frame = tk.Frame(chat_canvas, bg=PANEL)

canvas_window = chat_canvas.create_window(
    (0, 0),
    window=chat_frame,
    anchor="nw"
)

def update_chat(event=None):
    chat_canvas.configure(
        scrollregion=chat_canvas.bbox("all")
    )
    chat_canvas.itemconfigure(
        canvas_window,
        width=chat_canvas.winfo_width()
    )
    chat_canvas.yview_moveto(1.0)

chat_frame.bind("<Configure>", update_chat)
chat_canvas.bind("<Configure>", update_chat)

# ================= CHAT MESSAGE FUNCTION =================

def add_message(sender, message, is_bot=False, color=None):

    row = tk.Frame(chat_frame, bg=PANEL)
    row.pack(fill="x", pady=6)

    if is_bot:
        bubble_bg = CARD
        text_color = color if color else WHITE
        anchor = "w"
        padx = (0, 65)
        label_color = GREEN
    else:
        bubble_bg = "#123D2D"
        text_color = WHITE
        anchor = "e"
        padx = (65, 0)
        label_color = GREEN

    bubble = tk.Frame(
        row,
        bg=bubble_bg,
        padx=12,
        pady=9
    )
    bubble.pack(anchor=anchor, padx=padx)

    tk.Label(
        bubble,
        text=sender,
        font=("Segoe UI", 8, "bold"),
        fg=label_color,
        bg=bubble_bg
    ).pack(anchor="w")

    tk.Label(
        bubble,
        text=message,
        font=("Segoe UI", 10),
        fg=text_color,
        bg=bubble_bg,
        wraplength=480,
        justify="left",
        anchor="w"
    ).pack(anchor="w", pady=(5, 0))

    update_chat()

# ================= INITIAL BOT MESSAGE =================

add_message(
    "SNAPPFOOD BOT",
    "Hello! Welcome to SnappFood.\n\n"
    "We would love to hear your opinion. "
    "Send us your feedback and I'll analyze it.",
    is_bot=True
)

# ================= INPUT PANEL =================

input_panel = tk.Frame(
    main,
    bg=PANEL,
    padx=14,
    pady=11,
    highlightbackground=BORDER,
    highlightthickness=1
)
input_panel.pack(fill="x", pady=(12, 0))

tk.Label(
    input_panel,
    text="YOUR MESSAGE",
    font=("Segoe UI", 8, "bold"),
    fg=GREEN, bg=PANEL
).pack(anchor="w", pady=(0, 7))

feedback = tk.Text(
    input_panel,
    height=3,
    font=("Segoe UI", 10),
    bg=CARD,
    fg=WHITE,
    insertbackground=GREEN,
    relief="flat",
    padx=10,
    pady=9,
    wrap="word"
)
feedback.pack(fill="x", pady=(0, 9))

# ================= SUBMIT FUNCTION =================

def submit_feedback():

    ask = feedback.get("1.0", "end-1c").strip()

    if not ask:
        messagebox.showwarning(
            "Empty Message",
            "Please write your feedback first."
        )
        return

    # Display customer message in UI
    add_message(
        "YOU",
        ask,
        is_bot=False
    )

    # Analyze using existing NLP model
    vec_ask = vectorizer.transform([ask])
    result = model.predict(vec_ask)

    # Generate response
    if result[0] == 1:

        response = (
            "❤️ THANK YOU FOR YOUR FEEDBACK!\n\n"
            "خیلی ممنون از لطف شما\n\n"
            "We truly appreciate your positive feedback. "
            "We're happy that you enjoyed our product!\n\n"
            "Your satisfaction means a lot to us."
        )

        response_color = GREEN

    else:

        response = (
            "THANK YOU FOR SHARING YOUR OPINION.\n\n"
            "We appreciate your honesty and feedback.\n\n"
            "Your opinion helps us improve our product "
            "and provide a better experience."
        )

        response_color = WHITE

    # Display bot response in UI
    add_message(
        "SNAPPFOOD BOT",
        response,
        is_bot=True,
        color=response_color
    )

    # Clear input
    feedback.delete("1.0", "end")

# ================= SEND BUTTON =================

ttk.Button(
    input_panel,
    text="SEND MESSAGE  →",
    style="Modern.TButton",
    command=submit_feedback
).pack(anchor="e")

# ================= FOOTER =================

footer = tk.Frame(root, bg=BG)
footer.pack(fill="x", padx=28, pady=(0, 12))

tk.Label(
    footer,
    text="SNAPPFOOD AI | NLP SENTIMENT ANALYSIS",
    font=("Segoe UI", 7, "bold"),
    fg=GRAY, bg=BG
).pack(side="left")

tk.Label(
    footer,
    text="POWERED BY PARHAM SZF",
    font=("Segoe UI", 7, "bold"),
    fg=GREEN, bg=BG
).pack(side="right")

# ================= RUN =================

root.mainloop()
```
